***Step 0: Import***

```python
import cv2
import imutils
import matplotlib.pyplot as plt
import numpy as np
```

***Step 0.5: helper function***

```python
def show_image(img: np.ndarray, title: str = "") -> None:
    plt.imshow(img, cmap="gray")
    plt.title(title)
    plt.axis("off")
    plt.show()
```

***Step 1: Read the image***

```python
image = cv2.imread("tetris_blocks.png")
show_image(image, "Image")
```

***Step 2: Gray Scale***

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
show_image(gray, "Gray Image")
```

***Step 3: Thresholding***

```python
thresh = cv2.threshold(
    gray, 225, 255, cv2.THRESH_BINARY_INV
)[1]
show_image(thresh, "Thresholded Image")
```

***Step 4: Finding contours***

```python
cnts = cv2.findContours(
    thresh.copy(),
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

cnts = imutils.grab_contours(cnts)

output = image.copy()

cv2.drawContours(
        output, cnts, -1, (240, 0, 159), 5
    )
show_image(output, "Contours")
```

***Step 5: Finishing touch***

```python
text = f"I found {len(cnts)} objects!"

cv2.putText(
    output, text, (10, 25),
    cv2.FONT_HERSHEY_SIMPLEX,
    1, (240, 0, 159), 2
)

show_image(output, "Output")
```
