```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('PySparkExample.com').getOrCreate()
print(spark)
```

```python
scores = [("Nguyen Van A", 10.0), 
        ("Le Van Quy",9.5), 
        ("Tran Duc",8.0), 
        ("Pham Thi Nhung",9.), 
      ]
rdd = spark.sparkContext.parallelize(scores)
```

```python
df = rdd.toDF()
df.printSchema()
df.show(truncate=False)
```

```python
scoreColumns = ["name","score"]

df2 = rdd.toDF(scoreColumns)
```

```python
df2.printSchema()
df2.show(truncate=False)
```

```python
scores_DF = spark.createDataFrame(data=scores, schema = scoreColumns)
scores_DF.printSchema()
scores_DF.show(truncate=False)
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, DoubleType

# 1. Khởi tạo SparkSession (Cần thiết nếu bạn chạy script ở ngoài môi trường notebook có sẵn)
spark = SparkSession.builder.appName("ScoreApp").getOrCreate()

# 2. Khai báo dữ liệu đầu vào (tái tạo lại từ output trong ảnh)
scores = [
    ("Nguyen Van A", 10.0),
    ("Le Van Quy", 9.5),
    ("Tran Duc", 8.0),
    ("Pham Thi Nhung", 9.0)
]

# 3. Định nghĩa Schema MỚI (thay cho deptSchema) để khớp với dữ liệu điểm số
scoreSchema = StructType([
    StructField('name', StringType(), True),
    StructField('score', DoubleType(), True)
])

# 4. Khởi tạo DataFrame, truyền trực tiếp dữ liệu và schema vừa tạo
scoreDF = spark.createDataFrame(data=scores, schema=scoreSchema)

# 5. In ra cấu trúc và bảng dữ liệu
scoreDF.printSchema()
scoreDF.show(truncate=False)
```
