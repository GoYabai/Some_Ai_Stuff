```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.master("local[*]").getOrCreate()
sc = spark.sparkContext

# 1. Dữ liệu thô ban đầu
data = ["PySpark is fun", "learning PySpark is easy", "I love PySpark"]
rdd = sc.parallelize(data)

# 2. flatMap: Cắt từng câu thành các từ rời rạc
# Kết quả: ['PySpark', 'is', 'fun', 'learning', 'PySpark', 'is', 'easy', 'I', 'love', 'PySpark']
rdd_tu_vung = rdd.flatMap(lambda cau: cau.split(" "))

# 3. filter: Lọc bỏ những từ quá ngắn (ví dụ từ có 1 chữ như 'I')
rdd_loc = rdd_tu_vung.filter(lambda tu: len(tu) > 1)

# 4. map: Gắn cho mỗi từ một giá trị là 1 để tạo thành cặp (Key, Value)
# Kết quả: [('PySpark', 1), ('is', 1), ('fun', 1), ('learning', 1), ...]
rdd_cap_doi = rdd_loc.map(lambda tu: (tu, 1))

# 5. reduceByKey: Gom các từ giống nhau lại và cộng giá trị lên
# Logic: với 2 giá trị v1, v2 của cùng 1 từ, ta lấy v1 + v2
# Kết quả: [('PySpark', 3), ('is', 2), ('fun', 1), ('learning', 1), ('easy', 1), ('love', 1)]
rdd_dem_tu = rdd_cap_doi.reduceByKey(lambda v1, v2: v1 + v2)

# 6. sortByKey: Tùy chọn - Sắp xếp theo bảng chữ cái của tên từ (Key)
rdd_sap_xep = rdd_dem_tu.sortByKey(ascending=True)

# Gọi Action để xem kết quả
print(rdd_sap_xep.collect())
# Output: [('PySpark', 3), ('easy', 1), ('fun', 1), ('is', 2), ('learning', 1), ('love', 1)]
```
