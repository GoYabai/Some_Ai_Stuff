```python
import pyspark
from pyspark.sql import SparkSession

# khởi tạo
spark = SparkSession.builder \
    .appName('PySparkExample.com') \
    .getOrCreate()

print(spark)
```

```python
num_data = [1,2,3,4,5]

# chia mối phần tử trong list thành 1 RDD
num_rdd = spark.sparkContext.parallelize(num_data)
```

```python
print("RDD count :"+str(num_rdd.count()))
```

```python
text_data = [
    "Hello this is an example of Spark WordCount Example",
    "we are using PySpark",
    "PySpark is a great tool for Big Data Analysis"
]
text_rdd = spark.sparkContext.parallelize(text_data)

```

```python
print("RDD count :"+str(text_rdd.count()))
```

```python
# load file text
# Read per line and each RDD is a line in file
# Use when the file is large bcs safe
rdd2 = spark.sparkContext.textFile("/data/text_data.txt")
```

```python
# Load whole file
# Read whole file, each RDD is a key: file path, value: whole text
# Use when multiple small text file, easier to fail
rdd3 = spark.sparkContext.wholeTextFiles("/data/text_data.txt")
```
