```python
from pyspark import SparkConf, SparkContext
from pyspark.sql import SparkSession

# khởi tạo spark context là core của spark làm việc với RDD
conf = SparkConf().setAppName("Hoc_RDD").setMaster("local[*]")
sc = SparkContext(conf=conf)

# khởi tạo spark session hay là cả 1 chiếc xe bao gồm cả spark context làm việc với df, sql
spark = SparkSession.builder \
    .appName('PySparkExample.com') \
    .getOrCreate()

print(spark)
```
