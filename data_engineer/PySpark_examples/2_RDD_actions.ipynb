```python
import pyspark
from pyspark.sql import SparkSession

# khởi tạo
spark = SparkSession.builder.appName('PySparkExample.com').getOrCreate()
print(spark)
```

```python
num_data = [1,2,3,4,5]

# mỗi phần tử trong list thành 1 RDD
num_rdd = spark.sparkContext.parallelize(num_data)
```

```python
print("RDD count :"+str(num_rdd.count()))
```

```python
# lấy 3 RDD đầu tiên
print("RDD take :"+str(num_rdd.take(3)))
```

```python
# lấy 1 RDD đầu
print("RDD take :"+str(num_rdd.first()))
```

```python
# lấy 2 RDD cuối
print("RDD take :"+str(num_rdd.top(2)))
```

```python
# lấy RDD max
print("RDD take :"+str(num_rdd.max()))
```

```python
# lấy RDD min
print("RDD take :"+str(num_rdd.min()))
```
